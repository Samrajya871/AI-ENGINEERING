Mini Exercises

Create a database named shop.db with a products table containing:

id
name
price
quantity

Then:

Insert 5 products.
Display all products.
Update the price of one product.
Delete one product.
Display all remaining products.

```python
import sqlite3
try:
    connection=sqlite3.connect("shop.db")
    print("Database connected successfully!")
    cursor = connection.cursor()

except sqlite3.Error as e:
    print("DATABASE ERROR:",e)
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS products(
id INTEGER PRIMARY KEY,
name TEXT,
price INTEGER,
quantity INTEGER
)
""")

connection.commit()
```

```python
products_data = [
('Wireless Mouse', 29.99, 50),
('Mechanical Keyboard', 89.99, 35),
('Gaming Monitor', 249.50, 15),
('USB-C Hub', 45.00, 120),
('Bluetooth Headphones', 120.00, 25)
]

cursor.executemany("""
INSERT INTO products(name, price, quantity)
VALUES(?,?,?)
""", products_data)

connection.commit()
```

```python
cursor.execute("SELECT * FROM products")

products = cursor.fetchall()

for prod in products:
    print(prod)
```

```python
cursor.execute("UPDATE products SET price=100.00 WHERE id=2")
connection.commit()
```

```python
cursor.execute("DELETE FROM products WHERE id=3")
connection.commit()
```
