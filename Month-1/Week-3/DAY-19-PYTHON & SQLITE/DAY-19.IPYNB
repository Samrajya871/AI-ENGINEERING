```python
import sqlite3
```

```python
try:
    connection = sqlite3.connect("library.db")
    cursor=connection.cursor()

    cursor.execute("SELECT * FROM books")
except sqlite3.Error as e:
    print("DATABSE ERROR:",e)

finally:
    connection.close()
```

```python
cursor = connection.cursor()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS books(
id INTEGER PRIMARY KEY,
title TEXT,
author TEXT,
year INTEGER
)
""")

connection.commit()
```

Why commit()?

Changes remain temporary until committed.

Always commit after:

INSERT
UPDATE
DELETE
CREATE TABLE

```python
cursor.execute("""
INSERT INTO books(title, author, year)
VALUES(?, ?, ?)
""", ("Python Crash Course","Eric Matthes",2023))

connection.commit()
```

Why use ? placeholders?

This is called a parameterized query. It helps prevent SQL injection and is the recommended way to pass values into SQL statements.

```python
cursor.execute("SELECT * FROM books")

books = cursor.fetchall()

for book in books:
    print(book)

    #for only one row fetchone()
```

```python
cursor.execute("""
UPDATE books
SET year = ?
WHERE id=?
""",(2024,1))

connection.commit()
```

```python
cursor.execute("""
DELETE FROM books
WHERE id=?
""",(1,))

connection.commit()
```
