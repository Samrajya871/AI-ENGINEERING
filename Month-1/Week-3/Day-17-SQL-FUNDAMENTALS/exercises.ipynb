Mini Exercises

Create a table called books with these columns:

id
title
author
year

Then:

Insert 5 books.
Display all books.
Display only titles.
Find books published after 2020.
Sort books by year in descending order.
Display only the first 3 books.

#creating table
CREATE TABLE books(
   id INTEGER PRIMARY KEY,
   title TEXT,
   author TEXT,
   year INTEGER);


#TO INSERT 5 BOOKS
INSERT INTO books VALUES
(1, "ASOIAF","GRRM",1992),
(2,"LOTR","TOLKEIN",1934),
(3,"RR","PIERCE BROWN",2014),
(4,"DUNE","FRANK HERBET",1970),
(5,"METAMORPHOSIS","KRANZ KAFKA",1900);

#DISPLAY ALL BOOKS

SELECT * FROM books;

#display only title

SELECT * FROM books
WHERE year>2010;

#find books published after 2020
SELECT * FROM books
WHERE year>2020

#sort books by year in descending order
SELECT * FROM books
ORDER BY year DESC;

#DISPLAY ONLY FIRST 3 BOOKS
SELECT * FROM books
LIMIT 3;
