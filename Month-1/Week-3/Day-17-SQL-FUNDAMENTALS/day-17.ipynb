##SQL FUNDAMENTALS

What Is a Database?

Imagine storing 10,000 students in a Python list.

Possible? Yes.

Efficient? Not really.

Databases are designed to:

Store large amounts of data
Retrieve data quickly
Keep data organized
Prevent duplication
Support multiple users and applications

A Database Management System (DBMS) manages this data.


What Is SQL?

SQL = Structured Query Language

It's the standard language for interacting with relational databases.

Common SQL operations:

Create databases and tables
Insert records
Search records
Update records
Delete records

Think of SQL as the language you use to "talk" to a database.

Creating Your First Table

Open DB Browser for SQLite (recommended for beginners) or use the SQLite command-line tool.

Create a database named:

students.db

Create a table:

CREATE TABLE students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    course TEXT
);

Congratulations—you've created your first SQL table!


Inserting Data

Add some students:
INSERT INTO students VALUES
(1,'RAM',20,'AI');
INSERT INTO students VALUES
(2,'HARI',21,'DATA SCIENCE');


Reading Data

Display everything:

SELECT * FROM students;

Display only names:

SELECT name FROM students;

Display names and courses:

SELECT name, course FROM students;


Filtering with WHERE

Find students older than 20:

SELECT * FROM students
WHERE age > 20;

Find AI students:

SELECT * FROM students
WHERE course = 'AI';

Other comparison operators:

=
!=
<
>
<=
>=

Sorting Results

Ascending:

SELECT * FROM students
ORDER BY age ASC;

Descending:

SELECT * FROM students
ORDER BY age DESC;
