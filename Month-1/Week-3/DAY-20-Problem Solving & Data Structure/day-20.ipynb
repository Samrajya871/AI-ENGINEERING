BIG-O Notation:
It describes how an algorithm's runtime or memory usage grows as the input size increases.

common complexities

Complexity | Meaning           | Example
O(1)       | Constant          | Dictionary lookup
O(logn)    | Logarithmic       | Binary search
O(n)       | Linear            | Loop through a list
O(nlogn)   | Efficient sorting | Merge Sort
O(n^2)     | Quadratic         | Nseted loops
O(2^n)     | Exponential       | Some recursive algorithm

```python
#example

numbers = [5, 10, 15, 20]

for num in numbers:
    print(num)

#one loop over n times --> O(n)
```

```python
#arrays (python lists)

numbers = [10,20,30,40]

#common operations

numbers.append(50)
numbers.pop()
numbers.insert(1,15)
numbers.remove(30)
print(numbers[2])
```

Access By index O(1)
Append O(1) (amortized)
search O(n)
insert/delete in middle O(N)

```python
#PART 3 STRINGS
#SEQUENCES OF CHARACTERS

text= "Artificial Intelligence"

print(text.upper())
print(text.lower())
print(text.replace("Artificial", "Machine"))
print(text.split())
```

```python
so=""
for t in text:
    so = t + so
print(so)
```

```python
count=0
for t in text.lower():
    if t == "a":
        count+=1
    elif t=="e":
        count+=1
    elif t=="i":
        count+=1
    elif t=="o":
        count+=1
    elif t=="u":
        count+=1

print(count)
```
