```python
#constructot (init)
#Automatically runs whenever an object is created

class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
```

```python
#creating object
s1=Student("Samrajya",25)
print(s1.name)
print(s1.age)
```

```python
#methods

class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def introduce(self):
        print("Hi, I'm",self.name)


#creating object
s1=Student("Sam",25)
s1.introduce()
```
