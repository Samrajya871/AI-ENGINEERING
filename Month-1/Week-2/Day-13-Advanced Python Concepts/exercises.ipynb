Exercise 1

Create a lambda function that cubes a number.

```python
num=int(input("Enter the number: "))
cube=lambda x:x*x*x
print(cube(num))
```

Exercise 2

Use map() to convert:

[5,10,15]

into

[10,20,30]

```python
numbers=[5,10,15]

result=list(map(lambda x:x*2,numbers))

print(result)
```

Exercise 3

Use filter() to keep numbers greater than 50

```python
numbers=[1,4,5,50,51,75,77,3,56,78]

result=list(filter(lambda x: x>50,numbers))
print(result)
```

Exercise 4

Use reduce() to multiply

[2,3,4]

Expected

24

```python
from functools import reduce

numbers=[2,3,4]

total=reduce(lambda x,y:x*y,numbers)

print(total)

```

Exercise 5

Create

List comprehension
Dictionary comprehension
Set comprehension

using numbers from 1–10.

```python
#list comphrension

squares_list = [x**2 for x in range(1,11)]

print(squares_list)
```

```python
#dictionary comphrension
cubes_dict = {x: x**3 for x in range(1,11)}

print(cubes_dict)
```

```python
#set comphrension

remainders_set = {x %3 for x in range(1,11)}

print(remainders_set)
```
