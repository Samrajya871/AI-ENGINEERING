INHERITANCE
It allows one class to reuse another class's methods and attributes

```python
#Example

class Animal:
    def speak(self):
        print("Animal makes a sound")
    

class Dog(Animal):
    pass

d = Dog()
d.speak()
```

```python
#Example of adding new methods

class Animal:
    def eat(self):
        print("Eating....")
    

class Dog(Animal):
    def bark(self):
        print("woof")

d = Dog()
d.eat()
d.bark()
```

```python
#Mehod Overriding
#child classes can replace parent class methods

class Animal:
    def speak(self):
        print("Animal Sound")

class Dog(Animal):
    def speak(self):
        print("woof")


dog =Dog()
dog.speak()
```

```python
#super()
#sometimes we want to use parent's constructor


class Person:
    def __init__(self,name):
        self.name=name

class Student(Person):

    def __init__(self,name,roll):
        super().__init__(name)
        self.roll=roll


student= Student("Sam",20)
print(student.name)
print(student.roll)
```

#ENCAPUSLATION
#IT PROTECTS DATA

self._balance means internal use

self.__balance means better protection

```python
class Bank:
    def __init__(self,balance):
        self.__balance=balance

    

b=Bank(5000)
print(b._Bank__balance)
print(b.balance)
```

```python
#getters and setters
class Bank:
    def __init__(self,balance):
        self.__balance=balance

    def get_balance(self):
        return self.__balance

    def set_balance(self,amount):
        if amount>=0:
            self.__balance=amount
        else:
            print()

acc=Bank(5000)
acc.set_balance(-100)
acc.set_balance(6000)
print(acc.get_balance())
```

```python
#POLYMORPHISM

#different objects respond differently to the same method

class Dog:

    def speak(self):
        print("WOOF!")

class Cat:

    def speak(self):
        print("MEOW!")


animals=[Dog(), Cat()]

for animal in animals:
    animal.speak()
```

```python
#ABSTRACTION
#it hides unneccessary implementation details

#it uses the abc module

from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):
    def area(self):
        return 20

rec =Rectangle()
rec.area()
```

```python
#CLASS VARIABLES
#shared by all objects

class Student:
    school="ABC University"

    def __init__(self,name):
        self.name=name


s1=Student("Ram")
s2=Student("Hari")
print(s1.name,s1.school)
print(s2.name,s2.school)
```

```python
#class methods
#a specific type of method that is bound directly to the class itself rather
#than to any individual instance(object) of that class.

class Student:
    school="ABC"

    @classmethod
    def change_school(cls, new_name):
        cls.school=new_name


Student.change_school("XYZ University")
print(Student.school)
```

```python
#static methods
#no access to object or class

class Calculator:

    @staticmethod

    def add(a,b):
        return a+b


print(Calculator.add(5,7))

```

```python
#properties

class Person:

    def __init__(self,name):
        self._name=name

    @property

    def name(self):
        return self._name

person=Person("ram")
print(person.name)


```
