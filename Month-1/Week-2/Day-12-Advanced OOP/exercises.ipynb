Exercise 2

Create:

Vehicle

Child:

Bike
Car
Truck

Each overrides:

start()

```python
class Vehicle:
    def start(self):
        print("Vehicle is starting ")

class Bike(Vehicle):
    def start(self):
        print("Bike is starting ")

class Car(Vehicle):
    def start(self):
        print("Car is starting ")

class Truck(Vehicle):
    def start(self):
        print("Truck is starting ")


B=Bike()
C=Car()
t=Truck()
vehicle=[B, C, t]
for v in vehicle:
    v.start()




```

Exercise 3

Create a class

Employee

Private variable

__salary

Method

display_salary()

```python
class Employee():
    def __init__(self,salary):
        self.__salary=salary

    def display_salary(self):
        print(self.__salary)

e = Employee(5000)
e.display_salary() 
```

Exercise 4

Create a Calculator class using only static methods.

Methods:

add
subtract
multiply
divide

```python
class Calculator:

    @staticmethod
    def add(a,b):
        return a+b
    
    @staticmethod
    def sub(a,b):
        return a-b

    @staticmethod
    def mul(a,b):
        return a*b

    @staticmethod

    def div(a,b):
        return a/b
```

```python
print(Calculator.add(17,20))
```

Exercise 5

Create a Student class with a class variable

school_name

Change it using a class method.

```python
class Student:
    school_name="Adarsh"


    @classmethod
    def change_school(cls, new_name):
        cls.school_name=new_name

Student.change_school("Trinity")
print(Student.school_name)
```
