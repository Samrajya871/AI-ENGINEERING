```python
#exercise 1
#to create a class called book

class Book:
    def __init__(self,title,author):
        self.title=title
        self.author=author
    
    def display(self):
        print("Book Title:",self.title)
        print("Book Author:",self.author)

b1=Book("ASOIAF","GRRM")
b1.display()
```

```python
#exercise 2
#to create a class called car

class Car:
    def __init__(self,brand,model,year):
        self.brand=brand
        self.model=model
        self.year=year

    def show_details(self):
        print(f"The car brand is {self.brand} and the model is {self.model} and was launched in year {self.year}")


c1=Car("Toyota","Prius",2015)
c1.show_details()
```

```python
#Create three Student objects.
#Store them in a list.
#Loop through them.

class Student:
    def __init__(self,name):
        self.name=name

    def introduce(self):
        print(f"Hello, I am {self.name}.")

s1=Student("Ram")
s2=Student("Hari")
s3=Student("Sita")

student_list=[s1,s2,s3]

for student in student_list:
    student.introduce()
```

```python
#Create a bank account class

class BankAccount:
    def __init__(self,name,balance):
        self.name=name
        self.balance=balance
    
    def deposit(self,amount):
        self.balance+=amount
        return self.balance
    
    def withdraw(self,amount):
        if amount>self.balance:
            print("Amount is bigger than Balance")
        
        else:
            self.balance-=amount
            return self.balance
    
    def display_balance(self):
        print("Your total balance is",self.balance)

account=BankAccount("Ram",1000)
print("------ATM------")
print("1. Deposit")
print("2. Withdraw")
print("3. Display Balance")
print("4. Exit")
print("---------------")
while True:
    choice=int(input("Enter the choice: "))
    if choice==1:
        amount=float(input("Enter the amount u want to deposit: "))
        new_balance=account.deposit(amount)
    elif choice==2:
        amt=float(input("Enter the amount u want to withdraw:"))
        new_balance1=account.withdraw(amt)
    elif choice==3:
        account.display_balance()
    elif choice==4:
        print("Thanks for using")
        break
    else:
        print("Invalid choice")
```

```python
#Create a Movie Class

class Movie:
    def __init__(self,title,rating,director):
        self.title=title
        self.rating=rating
        self.director=director

    def display_movie(self):
        print(f"The movie name is {self.title} and has a rating of {self.rating} and the movie was directed by {self.director}. ")


m1=Movie("Interstellar",4.5,"Christopher Nolan")
m2=Movie("The Shining",4.7,"Stanley Kubrick")

m1.display_movie()
m2.display_movie()
```
